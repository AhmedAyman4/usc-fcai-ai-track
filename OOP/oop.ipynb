# Object-Oriented Programming (OOP) in Python

**What you'll learn:** how to organize your code using **classes** and **objects**, and the four pillars of OOP.

**Roadmap**
1. Classes & objects, `self`, constructors, class vs instance attributes
2. **Inheritance**: reuse code from another class
3. **Polymorphism**: the same action behaves differently for different objects
4. **Encapsulation**: protect your data
5. **Abstraction**: show only what matters


## 1. What is OOP?

**Object-Oriented Programming** is a way of organizing code around **objects** instead of loose functions and variables.

A real-life object, like a **car**, has:
- **Data** (called *attributes*): color, speed, brand
- **Actions** (called *methods*): drive, brake, honk

OOP lets us describe things like this in code, which makes programs **organized, reusable and easier to maintain**.

### Everything in Python is already an object!
Every data type you know (`int`, `str`, `list` ...) is a **class** that comes with its own methods.

````python
x = "ahmed"
print(type(x))             # <class 'str'>: str is a class, x is an object of it
print(x.upper())           # upper() is a method of the str class

numbers = [3, 1, 2]
print(type(numbers))
numbers.sort()             # sort() is a method of the list class
print(numbers)

# A peek at some methods that every string object has
print([m for m in dir(x) if not m.startswith("_")][:6])
````

## 2. Classes & Objects

- A **class** is a **blueprint** (like the plan of a house).
- An **object** (or *instance*) is a **real thing built from that blueprint** (the actual house). You can build many objects from one class.

Every object has three things:

| Part | Meaning | Example |
|---|---|---|
| **State** | its data (attributes) | name, age |
| **Behavior** | what it can do (methods) | introduce itself |
| **Identity** | a unique ID that makes it different from other objects | `id(obj)` |

Syntax: `class ClassName:` (we start class names with a **Capital letter**). We create an object by *calling* the class, and use the **dot** `.` to reach its attributes.

````python
class Student:
    pass                    # an empty class for now

s1 = Student()              # create an object
s1.name = "Ahmed"           # add attributes with the dot
s1.age = 20

s2 = Student()              # a second, independent object
s2.name = "Omar"
s2.age = 25

print(s1.name, s1.age)
print(s2.name, s2.age)

# Identity: every object has its own unique id
print(id(s1), id(s2))
print("Same object?", s1 is s2)
````

## 3. The `self` Parameter

Now let's give our class a **method** (a function inside a class).

`self` means **"the current object"**. When you write `s1.introduce()`, Python automatically passes `s1` as the first argument, so inside the method `self` *is* `s1`.

That's how `self.name` knows **whose** name to use. Without `self`, Python wouldn't know which object's data you mean.

> The first parameter of every normal method must be `self` (it's a strong convention to call it `self`).

````python
class Student:
    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old")

    def show_self(self):
        print("self has id:", id(self))

s1 = Student()
s1.name, s1.age = "Ahmed", 20
s2 = Student()
s2.name, s2.age = "Omar", 25

s1.introduce()      # self = s1
s2.introduce()      # self = s2

# Proof that self IS the object
s1.show_self()
print("s1 has id  :", id(s1))

# Behind the scenes, Python turns s1.introduce() into this:
Student.introduce(s1)
````

## 4. Constructor: `__init__`

Setting attributes one by one (like we did above) is long, and we might forget one.

The **constructor** is a special method called `__init__`. It runs **automatically** every time a new object is created, and its job is to set up the object's starting state.

```python
s1 = Student("Ahmed", 20)    # Python calls __init__ for us
```

> Technical note: Python first creates the object and then calls `__init__` to initialize it. Most people simply call `__init__` "the constructor".

If `__init__` needs arguments, you **must** provide them when creating the object.

````python
class Student:
    def __init__(self, name, age):
        print("A new student is being created...")   # runs automatically
        self.name = name        # save the value inside the object
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old")

s1 = Student("Ahmed", 20)
s1.introduce()

s2 = Student("Omar", 25)
s2.introduce()

# What if we forget the arguments?
try:
    s3 = Student()
except TypeError as e:
    print("Error:", e)
````

## 5. Constructor Types & "Overloading"

**1. Constructor with parameters:** what we just saw.

**2. Constructor with default values:** just like normal functions, parameters can have defaults. Then the arguments become optional.

**3. Constructor in inheritance:** uses `super()`. We'll see it after learning inheritance.

### Constructor overloading
In languages like Java you can write several constructors with different parameters. **Python doesn't support this**: if you write two `__init__` methods, the second one silently **replaces** the first.

The Python way is to use **default values** (or `*args` / `**kwargs`).

````python
class Student:
    def __init__(self, name="Guest", age=0):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old")

Student().introduce()                   # no arguments
Student("Ali").introduce()              # only the name
Student("Sara", 22).introduce()         # both
Student(age=19).introduce()             # keyword argument: choose which one to give
````

````python
# Proof that the second __init__ replaces the first one
class Demo:
    def __init__(self):
        print("first constructor")

    def __init__(self, x):
        print("second constructor, x =", x)

try:
    Demo()                 # the first constructor is gone
except TypeError as e:
    print("Error:", e)

Demo(5)
````

## 6. Class Attribute vs Instance Attribute

- **Instance attribute**: belongs to **one object** (created with `self.x = ...` inside `__init__`).
- **Class attribute**: written directly in the class body and **shared by all objects**.

If you assign to an object using the same name as a class attribute, Python creates a **new instance attribute** that hides the class one *for that object only*.

````python
class Student:
    school = "ABC School"            # class attribute (shared)

    def __init__(self, name):
        self.name = name             # instance attribute (one per object)

s1 = Student("Ahmed")
s2 = Student("Omar")
print(s1.school, "|", s2.school)     # both see the class value

s1.school = "XYZ School"             # creates an instance attribute on s1 only
print(s1.school, "|", s2.school, "|", Student.school)

Student.school = "New School"        # change the class attribute
print(s1.school, "|", s2.school)     # s1 kept its own value, s2 follows the class
````

# Inheritance

## 7. What is Inheritance?

**Inheritance** lets a **child** class (also called *subclass* or *derived class*) get all the attributes and methods of a **parent** class (*superclass* or *base class*).

Think "**is-a**": a Student **is a** Person, a Dog **is an** Animal.

**Why use it?**
- **Reuse code**: write common things once in the parent.
- **Model real hierarchies**: Person → Student, Employee, Teacher.
- **Easier maintenance**: fix the parent, and all children are fixed.

Syntax: `class Child(Parent):`

> Inheritance goes one way only: the child can use the parent's methods, but the parent can't use the child's.

````python
class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

    def show(self):
        print(f"Name: {self.name} | Age: {self.age} | City: {self.city}")

class Student(Person):          # Student inherits everything from Person
    def marks(self):
        print(f"{self.name}'s marks is 50")

s1 = Student("Ali", 25, "Cairo")
s1.show()                       # inherited from Person
s1.marks()                      # defined in Student

p1 = Person("Ahmed", 20, "Giza")
p1.show()
try:
    p1.marks()                  # a Person doesn't have marks()
except AttributeError as e:
    print("Error:", e)

print(isinstance(s1, Person))   # a Student is also a Person
````

## 8. The `super()` Function

What if the child needs **extra attributes** (like a student's faculty)? We give the child its own `__init__`, but that **replaces** the parent's one.

`super()` solves this: it lets the child **call the parent's version** first (to set `name`, `age`, `city`), then add its own stuff.

`super()` can also be used inside any method to reuse the parent's version of it.

````python
class Person:
    def __init__(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

    def show(self):
        print(f"Name: {self.name} | Age: {self.age} | City: {self.city}")

class Student(Person):
    def __init__(self, name, age, city, faculty):
        super().__init__(name, age, city)    # let Person set up the common data
        self.faculty = faculty               # then add what's new

    def show(self):
        super().show()                       # reuse the parent's show()
        print(f"Faculty: {self.faculty}")

s1 = Student("Ali", 25, "Cairo", "CS")
s1.show()
````

## 9. Types of Inheritance

| Type | Meaning |
|---|---|
| **Single** | one child, one parent |
| **Multilevel** | a chain: grandparent → parent → child |
| **Hierarchical** | one parent, several children |
| **Multiple** | one child with **several** parents |
| **Hybrid** | a mix of the types above |

When a class has several parents, Python searches for a method in a fixed order called the **MRO** (*Method Resolution Order*): first the class itself, then its parents from left to right.

````python
class Animal:
    def eat(self):
        print("eating")

class Dog(Animal):              # Single: Dog -> Animal
    def bark(self):
        print("woof")

class Puppy(Dog):               # Multilevel: Puppy -> Dog -> Animal
    def weep(self):
        print("whining")

class Cat(Animal):              # Hierarchical: Dog and Cat both inherit Animal
    def meow(self):
        print("meow")

p = Puppy()
p.eat(); p.bark(); p.weep()     # Puppy gets everything from the whole chain

# Multiple inheritance: one child, two parents
class Flyer:
    def move(self):
        print("flying")
    def fly(self):
        print("I can fly")

class Swimmer:
    def move(self):
        print("swimming")
    def swim(self):
        print("I can swim")

class Duck(Flyer, Swimmer):
    pass

d = Duck()
d.fly(); d.swim()
d.move()                                    # both parents have move(): who wins?
print([c.__name__ for c in Duck.mro()])     # the MRO: Flyer comes first
````

# Polymorphism

## 10. What is Polymorphism?

**Polymorphism** means **"many forms"**: the **same** method, function or operator behaves **differently** depending on the object it works with.

**Real-life example:** a remote control. The power button is the same interface, but it turns on a TV, an AC or a speaker differently.

**Why use it?** One consistent interface for different classes, so your code is more flexible and reusable, and easier to extend later.

We've already used it without knowing:

````python
# The same function len() works on different types
print(len("hello python"))
print(len([1, 2, 3, 4, 5]))
print(len({"a": 1, "b": 2}))

# The same operator + behaves differently
print(5 + 3)
print("py" + "thon")
print([1, 2] + [3, 4])
````

## 11. Method Overloading (same name, different parameters)

In Java or C++, you can write **several methods with the same name** but different parameters, and the right one is chosen *before* the program runs (*compile-time polymorphism*).

**Python doesn't support this**: a second method with the same name just replaces the first. Instead, we write **one** flexible method using **default values** or `*args` (which collects any number of arguments).

````python
class Demo:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):          # replaces the first add()
        return a + b + c

try:
    Demo().add(1, 2)
except TypeError as e:
    print("Error:", e)

print("-" * 30)

# The Python way: one flexible method using *numbers
class Calculator:
    def mul(self, *numbers):
        result = 1
        for n in numbers:
            result *= n
        return result

calc = Calculator()
print(calc.mul())
print(calc.mul(4))
print(calc.mul(4, 3))
print(calc.mul(4, 3, 5, 6))
````

## 12. Method Overriding (same name, different behavior)

This is the **main** polymorphism in Python (*runtime polymorphism*). A child class provides **its own version** of a method that already exists in the parent. Python decides **at runtime** which version to run, based on the object's real class.

````python
class Animal:
    def sound(self):
        return "some sound"

class Dog(Animal):
    def sound(self):                 # overrides Animal.sound
        return "Woof"

class Cat(Animal):
    def sound(self):                 # overrides Animal.sound
        return "Meow"

# Same call, different behavior
for animal in [Animal(), Dog(), Cat()]:
    print(type(animal).__name__, "says", animal.sound())
````

## 13. Polymorphism in Functions (Duck Typing)

A function doesn't need to care about an object's **class**, only that it **has the method it needs**. Python calls this *duck typing*: "if it walks like a duck and quacks like a duck, it's a duck".

The objects below are **unrelated** (no inheritance), yet one function works with all of them.

````python
class Pen:
    def use(self):
        return "writing"

class Eraser:
    def use(self):
        return "erasing"

def perform_task(tool):
    print(tool.use())        # works with ANY object that has use()

perform_task(Pen())
perform_task(Eraser())
````

# Encapsulation

## 14. What is Encapsulation? Access Specifiers

**Encapsulation** means bundling the data and the methods together in a class, and **controlling who can access the data**. It protects important data from being changed by mistake (like your bank balance).

Access levels are shown with underscores:

| Level | Syntax | Meaning |
|---|---|---|
| **Public** | `name` | accessible from anywhere (the default) |
| **Protected** | `_name` | for the class and its children only (a *convention*) |
| **Private** | `__name` | for the class only (Python renames it so you can't use it by accident) |

> ⚠️ Python is **not strict** like Java: protected and private are mostly *"please don't touch"* signs. Private names are renamed internally, which prevents accidents but is not real security.

````python
class Employee:
    def __init__(self, name, salary, password):
        self.name = name              # public
        self._salary = salary         # protected
        self.__password = password    # private

    def show_info(self):              # inside the class, everything is accessible
        print(f"{self.name} | {self._salary} | {self.__password}")

class Manager(Employee):
    def details(self):
        print(self.name, self._salary)        # child can use public + protected
        # print(self.__password)              # but NOT the private one

e1 = Employee("Ahmed", 50000, "1234")
e1.show_info()
print(e1.name)                # OK
print(e1._salary)             # works, but you shouldn't do this outside the class

try:
    print(e1.__password)      # private: blocked
except AttributeError as e:
    print("Error:", e)

Manager("Omar", 20000, "789").details()
````

## 15. Getters, Setters & Validation

The best use of encapsulation: keep the data **private**, and let other code change it **only through methods** that check the rules first.

- **Getter**: a method that *reads* the value.
- **Setter**: a method that *changes* the value after validating it.


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance          # private: can't be changed directly

    def get_balance(self):                # getter
        return self.__balance

    def deposit(self, amount):            # controlled change
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit must be positive")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid amount or not enough money")

acc = BankAccount("Ahmed", 1000)
acc.deposit(500)
acc.withdraw(2000)               # rejected
acc.withdraw(300)
print("Balance:", acc.get_balance())
````

# Data Abstraction

## 16. What is Abstraction?

**Abstraction** means showing only the **essential features** and hiding the complex details.

**Example:** with a smartphone you can call and take photos without knowing how signals or storage work.

**Why use it?**
- Users focus on **what** an object does, not **how**.
- Internal changes don't break the code that uses it.
- Makes complex systems easier to use.

### Abstraction vs Encapsulation (a common confusion)
| | Abstraction | Encapsulation |
|---|---|---|
| **Question** | *What* should be shown? | *How* is the data protected? |
| **Goal** | hide complexity | hide and protect data |
| **Tool in Python** | abstract classes (`ABC`) | `_protected`, `__private`, `@property` |

## 17. Abstract Base Classes (ABC)

In Python we achieve abstraction with the `abc` module:

- **Abstract class**: a blueprint for other classes. It **cannot** be used to create objects directly.
- **Abstract method** (`@abstractmethod`): a method with **no body**. Every child **must** implement it.
- **Concrete method**: a normal, fully written method that children can simply reuse.

If a child forgets to implement an abstract method, Python raises a `TypeError`. This **forces** all children to follow the same structure.

````python
from abc import ABC, abstractmethod

class Shape(ABC):                    # abstract class
    @abstractmethod
    def area(self):                  # abstract method: no body
        pass

    def describe(self):              # concrete method: shared by all shapes
        return f"I'm a {type(self).__name__} with area {self.area()}"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):                  # required implementation
        return round(3.14 * self.radius ** 2, 2)

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

for shape in [Circle(5), Rectangle(4, 3)]:
    print(shape.describe())

# 1) We can't create an object from an abstract class
try:
    Shape()
except TypeError as e:
    print("Error:", e)

# 2) A child that forgets the abstract method can't be created either
class Triangle(Shape):
    pass

try:
    Triangle()
except TypeError as e:
    print("Error:", e)
````
