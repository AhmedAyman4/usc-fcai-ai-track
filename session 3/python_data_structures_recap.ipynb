#  Part 1 – List

## 1.1 Quick Recap

* A **List** stores many items in one variable.
* Lists are **ordered** (items stay in their positions).
* Lists are **mutable** ✏️ → we **can** change them after creation.
* We access elements using an **index** (starting from `0`).
* We can **loop** through a list.

````python
students = ["Ahmed", "Mona", "Ali", "Sara"]

print(students[0])      # first element (index 0)
print(students[2])      # third element (index 2)

# Lists are mutable: we can change an element
students[1] = "Laila"
print(students)

# We can loop through a list
for name in students:
    print("Hello", name)
````

## 🧩 Problem 1 – Find the Highest Grade

Here is a list of grades:


**Task:** Find the highest grade **without using** the built-in `max()` function.

⏱️ Take a few minutes and try it yourself in the cell below.

````python
grades = [70, 85, 90, 60, 95, 88]

# Write your code here
highest_grade = grades[0] #70

for grade in grades:
    if grade > highest_grade:
        highest_grade = grade

print(highest_grade)
````

# Part 2 – Tuple

## 2.1 Quick Recap

* A **Tuple** also stores many items in one variable.
* Tuples are **ordered**.
* Tuples are **immutable** 🔒 → we **cannot** change them after creation.
* Tuples support **indexing**.
* We can **loop** through a tuple.
* We use **round brackets** `( )`.

````python
temperatures = (25, 27, 27, 30, 29, 27, 25)

print(temperatures[0])    # indexing works like a list
print(temperatures[3])

# Looping works too
for temp in temperatures:
    print(temp)
````

## 🧩 Problem 2 – Count a Specific Temperature

We recorded the temperature for several days:

❌ Do **not** use `.count()`.

````python
temperatures = (25, 27, 27, 30, 29, 27, 25)

# Write your code here
count = 0
for temp in temperatures:
    if temp == 27:
        count+=1
print (count)

# temperatures.count(27)
````

# 📖 Part 3 – Dictionary

## 3.1 Quick Recap

* A **Dictionary** stores data as **Key → Value** pairs.
* We use the **key** (not an index) to get the value.
* Dictionaries are **mutable** → we can change them.
* We can **loop** through the dictionary **keys**.
* We use **curly brackets** `{ }`.

````python
student = {
    "name": "Ahmed",
    "age": 20,
    "grade": [90,95,80,95,70]
}

print(student["name"])     # key "name"  →  value "Ahmed"
print(student["grade"])    # key "grade" →  value 90
````

> 🔑 **Remember:** In a list we use an **index** (`students[0]`). In a dictionary we use a **key** (`student["name"]`).

## 🧩 Problem 3 – Students Who Passed

Here are the grades of some students:

```python
grades = {
    "Ahmed": 80,
    "Mona": 95,
    "Ali": 65,
    "Sara": 90,
    "Omar": 55
}
```

**Task:** Print the **names** of students who scored **80 or more**.

Try it yourself below. ✍️

````python
grades = {
    "Ahmed": 80,
    "Mona": 95,
    "Ali": 65,
    "Sara": 90,
    "Omar": 55
}

# Write your code here

for student in grades:
    if grades[student]>= 80 :
        print(student)
````

### ✅ Solution

<details>

```python
    "Ali": 65,

    "Omar": 55
```

# 🎓 Final Recap

| Data Structure | Example | Mutable? | Access |
| -------------- | ------- | -------- | ------ |
| **List** | `["Ali", "Mona"]` | ✅ Yes | **Index** |
| **Tuple** | `("Ali", "Mona")` | ❌ No | **Index** |
| **Dictionary** | `{"name": "Ali"}` | ✅ Yes | **Key** |

### 🧠 Key ideas to remember

* **Mutable** = can be changed | **Immutable** = cannot be changed
* **Index** (lists & tuples) vs **Key** (dictionaries)
* Dictionaries store **Key → Value** pairs
* **Loop + Condition** solves many problems
