## 1. What is a Module?

A module is a Python file containing reusable code.

````python
import math
import math as m 

print(m.sqrt(16))
print(m.pi)

from math import sqrt, pi
print(sqrt(25))
````

## 2. Different Import Styles

````python
from math import sqrt, pi
print(sqrt(36))
print(pi)
````

## 3. Useful Standard Modules

````python
import random
import datetime

print(random.randint(1, 10))
print(datetime.date.today())
````

## 4. Packages & pip

A package is a collection of modules. `pip` is commonly used to install third-party Python packages.

````python
# Run this in a terminal, no
# t as normal Python code:
##python -m pip install requests
````

## 5. Virtual Environments

A virtual environment isolates project dependencies.

Typical commands:

```bash
python -m venv .venv
# Windows
.venv\\Scripts\\activate
# macOS/Linux
source .venv/bin/activate
```
