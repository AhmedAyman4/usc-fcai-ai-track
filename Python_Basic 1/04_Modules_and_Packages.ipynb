## 1. What is a Module?

A module is a Python file containing reusable code.

````python
import math

print(math.sqrt(25))
print(math.pi)
````

## 2. Different Import Styles

````python
from math import sqrt, pi
print(sqrt(36))
print(pi)
````

## 3. Useful Standard Modules

````python
import random
import datetime

print(random.randint(1, 10))
print(datetime.date.today())
````

## 4. Creating Your Own Module

Create a file named `calculator.py` next to this notebook with:

````python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
````

Then import it from another Python file or notebook:

````python
import calculator

print(calculator.add(2, 3))
print(calculator.multiply(4, 5))
````

## 5. Packages & pip

A package is a collection of modules. `pip` is commonly used to install third-party Python packages.

## 6. Virtual Environments

A virtual environment isolates project dependencies.

Typical commands:

```bash
python -m venv .venv
# Windows
.venv\\Scripts\\activate
# macOS/Linux
source .venv/bin/activate
```
