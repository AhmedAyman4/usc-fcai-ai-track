# 📋 Part 2 – List Comprehension

## 2.1 What is List Comprehension?

> **List Comprehension is a short way to create a new list using a loop.**

Let's start with the normal `for` loop we already know:

````python
numbers = [1, 2, 3, 4, 5]

squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
````

Now the **List Comprehension** version. Same result, just **one line**! ✨

````python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
````

### 🧩 The Syntax

```python
[expression for item in iterable]
```

| Part | Meaning | In our example |
|------|---------|----------------|
| **expression** | What we want to put in the new list | `number ** 2` |
| **item** | One element at a time | `number` |
| **iterable** | The list we loop through | `numbers` |

Read it like English: *"Give me `number ** 2` **for** each `number` **in** `numbers`."*

## 2.3 List Comprehension with `if`

What if we only want **some** items? We add a **condition**.

First, the normal loop:

````python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)

print(even_numbers)
````

Now the List Comprehension version:

````python
even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
````

### 🧩 The Syntax with a condition

```python
[expression for item in iterable if condition]
```

> 📌 The **condition goes at the end**, after the loop part.

# 🛠️ Part 3 – Functions

## 3.1 Why do we need Functions?

Look at this code:

````python
print("Hello Ahmed")
print("Hello Mona")
print("Hello Ali")
````

### 💡 Think

> **What if we want to do this 100 times? Or change the message?**

A **function** is a **reusable block of code** that performs a specific task.
We write it **once**, and use it **many times**. ♻️

## 3.2 Creating a Simple Function

````python
def say_hello():
    print("Hello!")
````

Nothing happened! Why? 🤔

### 🧩 Parts of a function

| Part | Meaning |
|------|---------|
| `def` | Tells Python "I am defining a function" |
| `say_hello` | The function **name** |
| `()` | Parentheses (needed for every function) |
| `:` | Colon at the end of the line |
| indentation | The code **inside** the function is indented |

> ⚠️ **Defining a function does NOT run it. Calling the function runs it.**

Now let's **call** it:

````python
say_hello()
say_hello()
````

## 3.3 Function with a Parameter

What if we want to say hello to **different names**? We give the function some **input**.

````python
def say_hello(name):
    print("Hello", name)

say_hello("Ahmed")
say_hello("Mona")
````

### 🧩 Parameter vs Argument

* `name` is the **parameter** → the variable inside the function definition.
* `"Ahmed"` is the **argument** → the real value we send when we call the function.

# 🔙 Part 4 – `return`

So far, our functions only **displayed** something. Sometimes we want the function to **give us a value back**.

````python
def add(a, b):
    return a + b
````

````python
result = add(5, 3)

print(result)
````

### 🧩 `print` vs `return`

| | What it does |
|---|---|
| `print()` | **Displays** something on the screen 👀 |
| `return` | **Sends a value back** from the function 📦 |

> With `return`, we can **store** the value in a variable (like `result`) and use it later.

### 💡 Think

What do you think happens if we use `print` instead of `return` in `add`?

<details>
<summary><b>Show answer</b></summary>

The function would show the number on the screen, but `result` would be empty (`None`), so we could not use the value later.

</details>

# 🎓 Final Recap

## List Comprehension

**Use it when:**

* You want to **create a new list**
* You have a **simple loop**
* You want **shorter and cleaner** code

```python
[expression for item in iterable]               # basic
[expression for item in iterable if condition]  # with condition
```

## Functions

**Use them when:**

* You have code you want to **reuse**
* You want to **organize** your code
* You want to give a task a **clear name**

```python
def function_name():                 # basic
    # code

def function_name(parameter):        # with parameter
    # code

def function_name(parameter):        # with return
    return result
```
