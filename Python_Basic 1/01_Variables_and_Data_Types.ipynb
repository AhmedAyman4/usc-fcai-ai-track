## 1. Variables

A variable is a name that refers to a value. Python creates the variable when you assign a value to it.

```python
name = "Ahmed"
age = 20


print(name)
print(age)
```

## 2. Basic Data Types

### Integer
Whole numbers.

```python
age = 20
print(type(age))
```

### Float
Numbers with a decimal part.

```python
price = 99.5
print(type(price))
```

### String
Text surrounded by quotes.

```python
city = "Cairo"
print(type(city))
```

### Boolean
A value that is either `True` or `False`.

```python
is_student = True
print(type(is_student))
```

## 3. Type Casting

Type casting means converting a value from one type to another.

```python
age_text = "21"
age = int(age_text)
height = float("170.5")
number_text = str(100)
print(age, height, number_text)
```

## 4. Input & Output

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
print(f"Hello {name}, you are {age} years old.")
print(f"Hello {name}, you are {age} years old.")
```
