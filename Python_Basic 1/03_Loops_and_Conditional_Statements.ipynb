#  03 — Loops & Conditional Statements

## Learning Objectives

- Use if, elif, and else.
- Repeat operations with for loops.
- Use while loops.
- Control loops with break and continue.
- Combine conditions and loops.

## 1. Conditional Statements

```python
age = 20

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

### Multiple Conditions

```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)
```

## 2. for Loop

```python
for number in range(1, 6):
    print(number)
```

### Looping through a List

```python
names = ["Ali", "Mona", "Omar"]
for name in names:
    print(f"Hello {name}")
```

## 3. while Loop

```python
counter = 1
while counter <= 5:
    print(counter)
    counter += 1
```

## 4. break & continue

```python
for number in range(1, 11):
    if number == 6:
        break
    print(number)
```

```python
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
```

## Sample Solutions

```python
for number in range(1, 21):
    if number % 2 == 0:
        print(number, "even")
```

```python
score = float(input("Score: "))
if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
```

```python
secret = 7
while True:
    guess = int(input("Guess: "))
    if guess == secret:
        print("Correct!")
        break
    print("Try again.")
```

```python
number = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

## Summary

Conditions make decisions. Loops repeat work. `break` stops a loop, while `continue` skips the current iteration.
