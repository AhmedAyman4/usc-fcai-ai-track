#  03 — Loops & Conditional Statements

## Learning Objectives

- Use if, elif, and else.
- Repeat operations with for loops.
- Use while loops.
- Control loops with break and continue.
- Combine conditions and loops.

## 1. Conditional Statements

```python
age = 20


if age >=18:
    print("Adult")
else:
    print("Minor")
```

```python
grade=int(input("Enter your grade: "))

if grade >=50:
    print("Pass")
else:
    print('fail')    
```

### Multiple Conditions

```python
score = 82

if score >= 90:        
    grade = "A"
elif score >= 80:
    grade = "B" 
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(grade)
```

## 2. for Loop

```python
list_of_numbers = [1, 2, 3, 4, 5]

for number in list_of_numbers:
   print(number)
```

### Looping through a List

```python
names = ["Ali", "Mona", "Omar"]


for name in names:
    print(f"Hello {name}")
```

## 3. while Loop

```python
counter = 1 
while counter <= 5:
    print(counter)
    counter += 1
```

## 4. break & continue

```python
for number in range(1, 11):
    ### 1 2 3 4 5 6
    if number == 6:
        break
    print(number)
    
for number in range(1, 11):
    if number == 6:
        continue
    print(number)    
    
```

```python
for number in range(1, 11):
    if number % 2 == 0:
        continue
    print(number)
```
