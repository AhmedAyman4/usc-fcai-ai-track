## 1. Lists

Lists are ordered and mutable collections.

```python
students = ["Ali", "Mona", "Omar"]
print(students)
print(students[0])
print(students[-1])
```

### Slicing

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[::2])
```

### Useful List Methods

```python
numbers = [3, 1, 4]
numbers.append(10)
numbers.insert(1, 99)
numbers.remove(4)
numbers.sort()
print(numbers)
```

## 2. Tuples

Tuples are ordered collections that cannot be changed after creation.

```python
point = (10, 20)
print(point[0])
print(type(point))
```

## 3. Dictionaries

A dictionary stores data as key-value pairs.

```python
student = {"name": "Mona", "age": 21, "gpa": 3.5}
print(student["name"])
print(student.get("city", "Unknown"))
student["city"] = "Cairo"
print(student)
```

## 4. Nested Structures

```python
students = [
    {"name": "Ali", "grade": 85},
    {"name": "Mona", "grade": 92},
    {"name": "Omar", "grade": 78}
]
print(students[1]["name"])
```

## 4.Sets

```python
# Creating a Set
numbers = {1, 2, 3, 4, 5}
print(numbers)

# Duplicate values are automatically removed
numbers = {1, 2, 2, 3, 3, 4}
print(numbers)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

# Union
print(A | B)

# Intersection
print(A & B)

# Difference
print(A - B)

# Symmetric Difference
print(A ^ B)
```

```python
numbers = {1, 2, 3}

numbers.add(4)
print(numbers)

numbers.remove(2)
print(numbers)

numbers.discard(10)  
print(numbers)
```
